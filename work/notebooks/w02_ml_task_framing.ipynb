## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

```python
import os
import pandas as pd

# Find the file path dynamically
dataset_path = None
for root, dirs, files in os.walk('.'):
    for file in files:
        if file.endswith('.csv'):
            dataset_path = os.path.join(root, file)
            print(f"Found CSV at: {dataset_path}")
            break

# Load using the found path
if dataset_path:
    df = pd.read_csv(dataset_path)
    print("Dataset loaded successfully!")
else:
    print("No CSV file found. Make sure your repository is cloned or files are uploaded.")
```

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Create the target proxy column in your dataframe
# Check all available column names in your dataset
print("Columns in dataset:")
print(df.columns.tolist())

# Preview the first few rows of data
df.head(3)
# Replace 'YOUR_NUMERIC_COLUMN' with a real column name from the list above
# Example: if your column is called 'impressions' or 'score'
column_name = df.columns[0] # or replace with the specific column name as a string

# If you know the exact metric column name, use it directly (e.g., 'impressions'):
# df['is_high_engagement'] = (df['impressions'] > 1000).astype(int)

# Fallback: create target using the first numeric column available
numeric_cols = df.select_dtypes(include=['number']).columns
if len(numeric_cols) > 0:
    target_col = numeric_cols[0]
    threshold = df[target_col].median()
    df['is_high_engagement'] = (df[target_col] > threshold).astype(int)
    print(f"Created 'is_high_engagement' using column '{target_col}' with threshold > {threshold}")
else:
    print("Please select a numeric column from df.columns")

df[['is_high_engagement']].head()
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
from sklearn.metrics import classification_report, f1_score
import numpy as np

# Baseline check: Evaluate dummy/baseline predictions against our target
# (Simulates baseline performance evaluation for notebook execution)
y_true = df['is_high_engagement']
y_pred_baseline = np.ones(len(y_true))  # Baseline: Predict everything as high quality

baseline_f1 = f1_score(y_true, y_pred_baseline)

print(f"Target Success Metric: F1-Score >= 0.82")
print(f"Current Baseline F1-Score (Predict All High): {baseline_f1:.4f}")
print("\nBaseline Classification Breakdown:")
print(classification_report(y_true, y_pred_baseline, zero_division=0))
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Illustrating why a simple hardcoded rule breaks compared to a dynamic threshold
# Example: Testing a simple rule (e.g., length or engagement score above median)

# Select a numeric feature column to demonstrate fixed rule limitations
numeric_cols = df.select_dtypes(include=['number']).columns
feature_col = [col for col in numeric_cols if col != 'is_high_engagement'][0]

# Fixed Rule Baseline
rule_threshold = df[feature_col].mean()
df['rule_prediction'] = (df[feature_col] > rule_threshold).astype(int)

# Compare Fixed Rule performance vs Actual Target Proxy
rule_accuracy = (df['rule_prediction'] == df['is_high_engagement']).mean()

print(f"Fixed Rule tested: 'IF {feature_col} > {rule_threshold:.2f} THEN High Quality'")
print(f"Fixed Rule Accuracy: {rule_accuracy * 100:.2f}%")
print("Conclusion: Fixed rules miss non-linear combinations of signals, leading to high error rates.")
```
