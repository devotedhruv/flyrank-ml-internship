## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import os, sys, subprocess
import pandas as pd

# Setup repository path for Colab environment
if "google.colab" in sys.modules:
    REPO_DIR = "flyrank-ml-internship"
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", "https://github.com/devotedhruv/flyrank-ml-internship", REPO_DIR], check=True)
    os.chdir(f"/content/{REPO_DIR}")

# Load raw starter dataset
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: Zero Impression Page Percentage
zero_imp_pct = (df['impressions_90d'] == 0).mean() * 100

# Number 2: Baseline Performance Concentration (Top Decile Clicks Share)
top_10_threshold = df['clicks_90d'].quantile(0.90)
top_10_click_share = (df[df['clicks_90d'] >= top_10_threshold]['clicks_90d'].sum() / df['clicks_90d'].sum()) * 100

# Number 3: Content Age Profile
median_age_days = df['content_age_days'].median()
over_1yr_pct = (df['content_age_days'] > 365).mean() * 100

print("--- Key Supporting Dataset Metrics ---")
print(f"1. Inactive/Zero-Impression Pages: {zero_imp_pct:.1f}%")
print(f"2. Clicks Captured by Top 10% Pages: {top_10_click_share:.1f}%")
print(f"3. Median Content Age: {median_age_days:.0f} days ({over_1yr_pct:.1f}% older than 1 year)")
```
